```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.events AS
SELECT
    TRIM(event_id) AS event_id,
    TRIM(user_id) AS user_id,
    TRIM(product_id) AS product_id,
    LOWER(TRIM(event_type)) AS event_type,
    CAST(event_timestamp AS TIMESTAMP) AS event_timestamp
FROM ecommerce.bronze.events
WHERE event_id IS NOT NULL
  AND user_id IS NOT NULL
  AND product_id IS NOT NULL
  AND event_timestamp IS NOT NULL;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.orders AS
SELECT
    TRIM(order_id) AS order_id,
    TRIM(user_id) AS user_id,
    CAST(order_date AS TIMESTAMP) AS order_date,
    LOWER(TRIM(order_status)) AS order_status,
    CAST(total_amount AS DECIMAL(18,2)) AS total_amount
FROM ecommerce.bronze.orders
WHERE order_id IS NOT NULL
  AND user_id IS NOT NULL
  AND order_date IS NOT NULL
  AND total_amount IS NOT NULL
  AND total_amount >= 0;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.order_items AS
SELECT
    TRIM(order_item_id) AS order_item_id,
    TRIM(order_id) AS order_id,
    TRIM(product_id) AS product_id,
    TRIM(user_id) AS user_id,
    CAST(quantity AS INT) AS quantity,
    CAST(item_price AS DECIMAL(18,2)) AS item_price,
    CAST(item_total AS DECIMAL(18,2)) AS item_total
FROM ecommerce.bronze.order_items
WHERE order_item_id IS NOT NULL
  AND order_id IS NOT NULL
  AND product_id IS NOT NULL
  AND user_id IS NOT NULL
  AND quantity > 0
  AND item_price >= 0
  AND item_total >= 0;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.products AS
SELECT
    TRIM(product_id) AS product_id,
    TRIM(product_name) AS product_name,
    INITCAP(TRIM(category)) AS category,
    TRIM(brand) AS brand,
    CAST(price AS DECIMAL(18,2)) AS price,
    CAST(rating AS DECIMAL(3,2)) AS rating
FROM ecommerce.bronze.products
WHERE product_id IS NOT NULL
  AND product_name IS NOT NULL
  AND price >= 0
  AND rating BETWEEN 0 AND 5;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.reviews AS
SELECT
    TRIM(review_id) AS review_id,
    TRIM(order_id) AS order_id,
    TRIM(product_id) AS product_id,
    TRIM(user_id) AS user_id,
    CAST(rating AS INT) AS rating,
    TRIM(review_text) AS review_text,
    CAST(review_date AS DATE) AS review_date
FROM ecommerce.bronze.reviews
WHERE review_id IS NOT NULL
  AND order_id IS NOT NULL
  AND product_id IS NOT NULL
  AND user_id IS NOT NULL
  AND rating BETWEEN 1 AND 5
  AND review_date IS NOT NULL;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.users AS
SELECT
    TRIM(user_id) AS user_id,
    TRIM(name) AS name,
    LOWER(TRIM(email)) AS email,
    INITCAP(TRIM(gender)) AS gender,
    TRIM(city) AS city,
    CAST(signup_date AS DATE) AS signup_date
FROM ecommerce.bronze.users
WHERE user_id IS NOT NULL
  AND name IS NOT NULL
  AND email IS NOT NULL
  AND signup_date IS NOT NULL;
```
