```python
%sql

CREATE OR REPLACE TABLE ecommerce.gold.customer_summary AS

SELECT
    u.user_id,
    u.name,
    u.email,
    u.gender,
    u.city,
    u.signup_date,

    COUNT(DISTINCT o.order_id) AS total_orders,

    COALESCE(SUM(o.total_amount), 0) AS total_spend,

    COALESCE(AVG(o.total_amount), 0) AS average_order_value,

    MIN(o.order_date) AS first_order_date,

    MAX(o.order_date) AS last_order_date,

    DATEDIFF(
        CURRENT_DATE(),
        u.signup_date
    ) AS customer_days_since_signup

FROM ecommerce.silver.users u

LEFT JOIN ecommerce.silver.orders o
    ON u.user_id = o.user_id

GROUP BY
    u.user_id,
    u.name,
    u.email,
    u.gender,
    u.city,
    u.signup_date;
```
