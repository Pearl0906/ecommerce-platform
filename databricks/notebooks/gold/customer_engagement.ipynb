```python
%sql

CREATE OR REPLACE TABLE ecommerce.gold.customer_engagement AS

SELECT
    u.user_id,
    u.name,
    u.city,

    COUNT(e.event_id) AS total_events,

    COUNT(
        CASE
            WHEN e.event_type = 'view'
            THEN 1
        END
    ) AS product_views,

    COUNT(
        CASE
            WHEN e.event_type = 'click'
            THEN 1
        END
    ) AS product_clicks,

    COUNT(
        CASE
            WHEN e.event_type = 'cart'
            THEN 1
        END
    ) AS cart_events,

    COUNT(
        CASE
            WHEN e.event_type = 'purchase'
            THEN 1
        END
    ) AS purchase_events,

    MIN(e.event_timestamp) AS first_event_timestamp,

    MAX(e.event_timestamp) AS last_event_timestamp

FROM ecommerce.silver.users u

LEFT JOIN ecommerce.silver.events e
    ON u.user_id = e.user_id

GROUP BY
    u.user_id,
    u.name,
    u.city;
```
