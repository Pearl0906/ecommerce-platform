```python
%sql

CREATE OR REPLACE TABLE ecommerce.gold.sales_summary AS

SELECT
    CAST(order_date AS DATE) AS sales_date,

    COUNT(DISTINCT order_id) AS total_orders,

    COUNT(DISTINCT user_id) AS unique_customers,

    SUM(total_amount) AS total_revenue,

    AVG(total_amount) AS average_order_value,

    MIN(total_amount) AS minimum_order_value,

    MAX(total_amount) AS maximum_order_value

FROM ecommerce.silver.orders

WHERE order_status NOT IN ('cancelled', 'canceled')

GROUP BY
    CAST(order_date AS DATE);
```
