```python
%sql

CREATE OR REPLACE TABLE ecommerce.gold.product_performance AS

SELECT
    p.product_id,
    p.product_name,
    p.category,
    p.brand,
    p.price,
    p.rating,

    COUNT(DISTINCT oi.order_id) AS total_orders,

    SUM(oi.quantity) AS units_sold,

    SUM(oi.item_total) AS total_revenue,

    AVG(oi.item_price) AS average_selling_price,

    AVG(r.rating) AS average_review_rating,

    COUNT(DISTINCT r.review_id) AS total_reviews

FROM ecommerce.silver.products p

LEFT JOIN ecommerce.silver.order_items oi
    ON p.product_id = oi.product_id

LEFT JOIN ecommerce.silver.reviews r
    ON p.product_id = r.product_id

GROUP BY
    p.product_id,
    p.product_name,
    p.category,
    p.brand,
    p.price,
    p.rating;
```
